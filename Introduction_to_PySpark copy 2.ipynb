```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("PySpark Essential Training") \
    .master("local[*]") \
    .getOrCreate()
```

```python
taxi = spark.read.parquet("C:/Users/Jabulani.Mcineka/workspace/PySpark-Essential-Training/data/yellow_tripdata_2025-01.parquet")
```

```python
taxi.createOrReplaceTempView("taxi")
```

```python
spark.sql("SELECT * FROM taxi where total_amount > 50").show()
```

```python
spark.sql("SELECT * FROM taxi where total_amount > 50")\
     .filter("passenger_count > 2")\
     .select("payment_type", "passenger_count", "total_amount")\
     .show()
```

```python
query = '''
select payment_type, passenger_count, total_amount from taxi where total_amount > 50 and passenger_count > 2
'''
```

```python
spark.sql(query).show()
```
