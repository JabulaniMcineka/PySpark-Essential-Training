```python
avg_fare =df.groupBy('payment_type').agg(avg('total_amount')).sort('payment_type')
```

```python
# Write to CSV using PySpark by collecting the data and using local I/O
import os

csv_file = "C:/Users/Jabulani.Mcineka/workspace/PySpark-Essential-Training/avg_fare.csv"

# Remove the existing file/directory if it exists
if os.path.exists(csv_file):
    if os.path.isdir(csv_file):
        import shutil
        shutil.rmtree(csv_file)
    else:
        os.remove(csv_file)

# Use PySpark to compute and collect the data
# This leverages PySpark's distributed processing capabilities
data = avg_fare.collect()
columns = avg_fare.columns

# Write to CSV using direct file I/O
with open(csv_file, 'w') as f:
    # Write header
    f.write(','.join(columns) + '\n')
    # Write data rows
    for row in data:
        values = [str(v) if v is not None else '' for v in row]
        f.write(','.join(values) + '\n')

print(f"Successfully wrote to {csv_file}")
print(f"Columns: {columns}")
print(f"Rows written: {len(data)}")
```
